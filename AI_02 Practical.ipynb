```python
map_rooms = {
    "A": ["B", "C"],
    "B": ["D"],
    "C": ["D"],
    "D": ["G"],
    "G": []
}

print("Our map:")
print("        A")
print("       / \\")
print("      B   C")
print("       \\ /")
print("        D")
print("        |")
print("        G (Treasure 💎)")

print("Doors from each room:", map_rooms)
```

```python
def bfs(start, goal):
    to_do = [start]
    visited = []
    order = []

    while to_do:
        room = to_do.pop(0)

        if room in visited:
            continue

        visited.append(room)
        order.append(room)

        if room == goal:
            return order

        for nxt in map_rooms[room]:
            to_do.append(nxt)

    return order


bfs_order = bfs("A", "G")

print("BFS checked the rooms in the order:", bfs_order)
print("Number of rooms BFS checked:", len(bfs_order))
```

```python
def dfs(start, goal):
    to_do = [start]
    visited = []
    order = []

    while to_do:
        room = to_do.pop()

        if room in visited:
            continue

        visited.append(room)
        order.append(room)

        if room == goal:
            return order

        for nxt in map_rooms[room]:
            to_do.append(nxt)

    return order


dfs_order = dfs("A", "G")

print("DFS checked the rooms in the order:", dfs_order)
print("Number of rooms DFS checked:", len(dfs_order))
```

```python
door_cost = {
    ("A", "B"): 1,
    ("B", "D"): 1,
    ("A", "C"): 5,
    ("C", "D"): 1,
    ("D", "G"): 1
}


def path_cost(path):
    total = 0

    for i in range(len(path) - 1):
        door = (path[i], path[i + 1])
        total += door_cost[door]

    return total


path1 = ["A", "B", "D", "G"]
path2 = ["A", "C", "D", "G"]

print("Path 1 A -> B -> D -> G costs:",
      path_cost(path1))

print("Path 2 A -> C -> D -> G costs:",
      path_cost(path2))
```
